Este notebook permite visualizar y analizar las redes de co-ocurrencia de hashtags a partir de los datos extraídos con `InstagramScraperV2`. En este ejemplo, analizamos el nicho de **#astrophotography**.

```python
# 1. Importación de la clase actualizada
from InstagramGraph import InstagramGraph
import pandas as pd
import nest_asyncio

# Habilitar soporte para entornos asíncronos si es necesario
nest_asyncio.apply()
```

## Ejemplo de Análisis con Datos de Astrofotografía

Cargamos el archivo CSV generado por el scraper para el hashtag #astrophotography.

```python
# Ruta al archivo extraído
csv_file = "data/scrape_results_astrophotography.csv" 

# Inicializar el grafo. 
# Por defecto usa: post_caption, user_name y post_hashtags
ig = InstagramGraph(csv=csv_file)

# Generar características (procesamiento de texto y hashtags)
ig.getFeatures(translate=False, lemma=False)
```

```python
# Seleccionar datos
# max_posts: filtra usuarios que publican demasiado (posibles bots/spam)
ig.selectData(max_posts=10, lemma=False)

# Construir el grafo de co-ocurrencia
# min_frequency: solo muestra conexiones que aparecen al menos N veces
ig.buildGraph(min_frequency=0)
```

```python
# Visualización interactiva de la red
ig.plotGraph(sizing=10, node_size='adjacency_frequency', community_plot=True)
```

```python
# Visualización de comunidades (Sunburst)
ig.plotCommunity()
```

```python
# Guardar resultados
ig.savePlot(plot='map')
ig.saveTables()
```
